```python
import numpy as np
import matplotlib.pyplot as plt
import os
import sys

util_dir = os.getcwd()+ "/utils"
model_dir =  os.getcwd()+ "/models/"
path = model_dir + 'ECG_Echo_CNN/'
```

```python
from scipy.ndimage import zoom

def upsample_cam(cam, target_length=5000):
 
    n_leads, T = cam.shape
    factor = target_length / T
    cam_up = zoom(cam, (1, factor), order=1)   
    return cam_up
```

```python
def plot_gradcam_all_leads(cam, figsize=(20, 12)):
    n_leads = cam.shape[0]

    plt.figure(figsize=figsize)

    for i in range(n_leads):
        plt.subplot(n_leads, 1, i+1)
        plt.imshow(cam[i:i+1, :], aspect='auto', cmap='jet')
        plt.ylabel(f"L{i+1}", rotation=0, labelpad=20)

        if i == 0:
            plt.title("Grad-CAM Heatmaps for All 12 Leads")

        plt.xticks([])
        plt.yticks([])

    plt.xlabel("Time (downsampled)")
    plt.tight_layout()
    plt.show()
```

```python
def plot_gradcam_overlay(ecg, cam_up, y_sample, figsize=(22, 18)):
    time = np.arange(ecg.shape[1])
    n_leads = ecg.shape[0]

    plt.figure(figsize=figsize)

    for i in range(n_leads):
        plt.subplot(n_leads, 1, i+1)

        # Plot ECG waveform
        plt.plot(time, ecg[i,:,0], color='black', linewidth=0.7)

        # Grad-CAM overlay
        plt.imshow(
            cam_up[i:i+1, :],         # single row heatmap
            extent=[0, len(time), ecg[i,:,0].min(), ecg[i,:,0].max()],
            cmap='jet',
            alpha=0.45,
            aspect='auto'
        )

        plt.title(f"Lead {i+1}   (y = {y_sample})")
        plt.ylabel(f"L{i+1}", rotation=0, labelpad=20)
        plt.xticks([])

        if i == n_leads - 1:
            plt.xlabel("Time")

    plt.tight_layout()
    plt.show()
```

```python
x_sample = np.load(path + 'cam_1_x.npy',allow_pickle=True)
y_sample = np.load(path + 'cam_1_y.npy',allow_pickle=True)
cam = np.load(path + 'cam_1.npy',allow_pickle=True)
cam_up = upsample_cam(cam, target_length=5000)
plot_gradcam_overlay(x_sample, cam_up, y_sample)
```
