```python
import numpy as np
import tensorflow as tf
import os
```

```python
model_dir = os.getcwd()+ "/models/"
ext_path = os.getcwd()+ "/data/EchoNext/"
model_file = model_dir + 'ECG_Echo_CNN/model.h5'

model = tf.keras.models.load_model(model_file)
print("Model loaded:", model_file)
```

```python
model.summary()
```

```python
# load EchoNext sample

x_train = np.load(ext_path + 'x_train.npy',allow_pickle=True)
x_test  = np.load(ext_path + 'x_test.npy',allow_pickle=True)
y_train = np.load(ext_path + 'y_train.npy',allow_pickle=True)
y_test  = np.load(ext_path + 'y_test.npy',allow_pickle=True)
```

```python
y_pred_train= model.predict(x_train)
np.save(ext_path+'y_pred_train.npy',np.array(y_pred_train))
```

```python
y_pred_test= model.predict(x_test)
np.save(ext_path+'y_pred_test.npy',np.array(y_pred_test))
```
